# 🧩 Reto 1 – Temperatura
**Objetivo:** clasificar la temperatura del agua y evaluar posible estrés térmico.

**Datos:** `../data/muestras_calidad_agua.csv`

### Pasos
1. Carga el CSV en `df`.
2. Extrae `sitio` y `temperatura` en `df_temp`.
3. Crea `categoria_temp` con np.where o .loc.
4. Crea `riesgo_termico`.
5. Filtra a `df_caliente` (alta temperatura).
6. Exporta `resultado_temperatura.csv`.

> **Reto adicional:** porcentaje de sitios con riesgo térmico.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/muestras_calidad_agua.csv")
```

```python
df_temp = df[["sitio", "temperatura"]]
```

```python
df.head()
```

```python
df_temp
```

```python
df_temp["categoria_temp"] = np.where(
    df_temp["temperatura"] > 25,
    "Alta",
    "Normal"
)
```

```python
df_temp.head()
```

```python
df_temp["riesgo_termico"] = np.where(
    df_temp["temperatura"] > 28,
    "Sí",
    "No"
)
```

```python
df_temp.head()
```

```python
df_caliente = df_temp[df_temp["categoria_temp"] == "Alta"]
```

```python
df_caliente
```

```python
df_temp.to_csv("../data/resultado_temperatura.csv", index=False)
```

```python
porcentaje = (df_temp["riesgo_termico"] == "Sí").mean() * 100

print("Porcentaje de sitios con riesgo térmico:", porcentaje, "%")
```

El análisis de temperatura permitió clasificar los sitios según sus condiciones térmicas e identificar posibles riesgos asociados al aumento de la temperatura del agua. Mediante el uso de estructuras condicionales y funciones de NumPy fue posible categorizar los registros y detectar sitios con potencial estrés térmico.

Los resultados muestran que algunos cuerpos de agua presentan temperaturas elevadas, lo que podría afectar los organismos acuáticos y alterar el equilibrio ecológico. Además, el cálculo del porcentaje de sitios con riesgo térmico facilitó una mejor interpretación de la situación ambiental evaluada.
